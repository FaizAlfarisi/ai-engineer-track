## M3 — Palindrome Check

**Rule:** A palindrome reads the same forwards and backwards (e.g. `"radar"`). For each word below, print whether it is a palindrome, ignoring case.

**Words:** `"radar"`, `"Racecar"`, `"hello"`, `"madam"`

**Output format:** `radar: True` (one line per word)

# Challenge — Text Processor CLI

**Goal:** read a paragraph of text and print a processing report.

**Rule (all features are REQUIRED):**
1. Prompt for a paragraph (may be multi-line).
2. Prompt for a `target` word.
3. Print: word count, character count (excluding spaces).
4. Print the full paragraph reversed (all characters).
5. Print how many times `target` appears (case-insensitive).
6. Print the paragraph with every occurrence of `target` replaced by `***` (case-insensitive).
7. Print the MOST COMMON word and its count (case-insensitive, punctuation ignored — assume words are separated by spaces; ties go to the first-seen word).

**Output format:**
```
=== TEXT PROCESSOR ===
words: 8
chars (no spaces): 39
reversed: ...
'ai' count: 2
redacted: ...
most common: 'the' (3 times)
```

**Constraints:** use `split`, `join`, `lower`, `replace`, `count`; loops allowed; no functions needed. Keep the section headers exactly as above.

````python
prompt = """
The system will parse prompt to execute the script. However, since prompt 
contains an uppercase string, it creates a mismatch with the target function. If 
engineers process PROMPT without sanitizing it first, the runtime environment 
will reject the input. Therefore, we must never pass prompt directly into the 
compiler; instead, we should reformat prompt before deployment.
"""
print(f"words: {len(prompt.split())}")
print(f"chars (no spaces): {len(prompt.replace(" ", ""))}")
print(f"reversed: {prompt[::-1]}")
target = input("What is the target word?").lower()
print(f"{target} count: {prompt.lower().count(target)}")
print(f"redacted: {prompt.replace(target.lower(), "***").replace(target.upper(), "***").replace(target.title(), "***")}")
cleaned_prompt = prompt.lower().split()
most_common_word = max(cleaned_prompt, key=cleaned_prompt.count)
count = cleaned_prompt.count(most_common_word)
print(f"most common: '{most_common_word}' (appears {count} times)")
````
