## E2 — Tuple unpacking swap

**Task:** `a, b = 3, 7`. Swap them **without a temp variable** and print
`a b`. Then define `min_max(nums)` that returns a tuple of `(min, max)`
and unpack its result into `lo, hi`.

**Output:**
```
7 3
2 9
```

**Constraint:** no temporary variable for the swap. `min_max` must `return
(min(nums), max(nums))` and must be called with `min_max([4, 9, 2])`.

````python
a, b = 3, 7
b, a = a, b
print(a, b)

def min_max(nums):
    return (min(nums), max(nums))

lo, hi = min_max([4, 9, 2])
print(lo, hi)
````

## E4 — Counting with `dict.get`

**Task:** count the letters in `word = "banana"` using `counts[ch] =
counts.get(ch, 0) + 1` inside a loop. Print the dict.

**Output:**
```
{'b': 1, 'a': 3, 'n': 2}
```

**Constraint:** the loop body must use `.get(ch, 0)` — no `if ch in counts`
check, no manual default. Word is exactly `"banana"`, in order.

````python
counts = {}
word = "banana"

for ch in word.lower():
    counts[ch] = counts.get(ch, 0) + 1

print(counts)
````

## M2 — Grouping with `setdefault`

**Task:** `students = [("ada", "eng"), ("ben", "ai"), ("cyn", "eng"),
("dan", "ai"), ("ela", "ai")]`. Build a dict mapping each department to
a list of its students using **`setdefault`**. Print the dict.

**Constraint:** the loop must use `groups.setdefault(dept, []).append(name)`

## H2 — Interview drill: two-sum

Given a list of numbers and a target, return the pair of numbers from the
list that **add up to the target**, or `None` if no such pair exists.

```python
print(two_sum([3, 5, 2, 7], 10))   # 3 + 7
print(two_sum([1, 2, 3], 9))
```

**Output:**
```
(3, 7)
None
```

**Constraints (interview-grade):**
- **Single pass** — no nested loops, no `in list` membership. This is
  O(n), not O(n²).
- Use a **set** of numbers seen so far: for each `n`, check whether
  `target - n` is already in the set (O(1)); if yes, return that pair,
  otherwise add `n` to the set.
- Return the pair `(smaller, larger)` order as shown.
- No `return` before the loop ends, except the `None` fallback.

````python
def two_sum(numbers: list, target: int):
    seen = set()
    result = None
    for num in numbers:
        residual = target - num
    
        if residual in seen:
            if residual < num:
                result = (residual, num)
            else:
                result = (num, residual) 
            break
    
        seen.add(num)
    return result
    
print(two_sum([3,5,2,7], 10))
print(two_sum([1,2,3], 9))
````

## Challenge — Store Inventory CLI

Build a small inventory manager as a **function-decomposed CLI** — the
hardest thing so far. Everything runs off one dict: `products` maps a
product **name** to a dict of `{"price": float, "qty": int,
"category": str}`.

### Commands (lines of input, space-separated)

```
add NAME PRICE QTY CATEGORY    # insert a new product
restock NAME QTY               # add QTY to existing stock
sell NAME QTY                  # subtract QTY if stock allows
report                         # print the summary below
quit                           # end the program
```

### Behavior rules

- `add` of an existing name is **rejected**: print `exists: NAME`.
- `add` validates: PRICE and QTY must be numeric, QTY must be a whole
  number. If invalid, print `invalid: NAME` and do not insert.
- `sell` with QTY > stock is rejected: print `insufficient: NAME has N`.
- `sell`/`restock` of an unknown name: print `unknown: NAME`.
- Print `added: NAME` / `restocked: NAME -> N` / `sold: NAME, left N`
  on success.

### report output (exact format)

```
=== inventory ===
apple: 16 x 1.50 = 24.00 (fruit)
banana: 3 x 0.50 = 1.50 (fruit)
total value: 25.50
categories: fruit
low stock: banana
```

- One line per product (insertion order), `price: {price:.2f}` and
  `= {qty * price:.2f}`.
- `total value:` is the sum of `qty * price` across all products
  (`{:.2f}`).
- `categories:` lists distinct categories **sorted**, joined by space.
  Build it from a **set**. If empty, print `categories: none`.
- `low stock:` lists products with `qty < 5`, comma-separated. If none,
  print `low stock: none`.

### Example session (input → output)

```
add apple 1.5 10 fruit     → added: apple
add banana 0.5 3 fruit     → added: banana
add apple 2.5 5 fruit      → exists: apple
sell apple 4               → sold: apple, left 6
sell apple 99              → insufficient: apple has 6
restock apple 10           → restocked: apple -> 16
add tea 2.0 8 drinks       → added: tea
report                     → (summary above, plus tea line)
quit                       → (program ends)
```

### Structure constraints (this is the real test)

Decompose into **functions**, each doing one job:

```python
def add_product(products, name, price, qty, category): ...  # -> str message
def restock(products, name, qty): ...                       # -> str message
def sell(products, name, qty): ...                          # -> str message
def report(products): ...                                   # -> prints summary
def main(): ...                                             # -> loops, calls input()
```

- Every command handler **returns a string**; only `main()` and `report()`
  print. No printing inside handlers (testable by calling the function and
  checking the string).
- `main()` parses the command word and routes, then `print(...)`s the
  handler's return value.
- All user input flows through `input()` in `main()` — handlers take
  already-parsed values.
- `report()` builds the categories list from a set, and the low-stock list
  by scanning `products.items()`.

Verify by running the example session line-for-line and matching every
printed line.

````python
def add_product(products: dict, name, price, qty, category):
    if name in products:
        return f"exists: {name}"
    else:
        products[name] = {
            "price": price,
            "qty": qty,
            "category": category
        }
        return 

def restock(products: dict, name, qty):
    if name in products:
        products[name]['qty'] += qty
        return 
    else:
        return f"unknown: {name}"

def sell(products, name, qty):
    if name in products:
        if qty > products[name]['qty']:
            return f"insufficient: {name} has {products[name]['qty']}"
        else:
            products[name]['qty'] -= qty
        return 
    else:
        return f"unknown: {name}"

def report(products: dict):
    print("=== inventory ===")
    total_value = 0
    categories = set()
    for name in products:
        qty = products[name]['qty']
        price = products[name]['price']
        category = products[name]['category']
        categories.add(category)
        value = qty * price
        total_value += value
        
        print(f"{name}: {qty} x {price:.2f} = {value:.2f} ({category})")
        
    print(f"total value: {total_value:.2f}")
    print(f"categories: {" ".join(sorted(categories))}")
    
    low_stock = [name for name, product in products.items() if product['qty'] < 5]
    print(f"low_stock: {" ".join(low_stock)}")

def main():
    products = {}
    while True:
        command = tuple(input().strip().lower().split())
        
        match command[0]:
            case "add":
                name, price, qty, category = command[1:]
                if not price.replace(".", "").isdigit() or float(price) <= 0 or not qty.isdigit() or int(qty) <= 0:
                    print(f"invalid: {name}")
                    continue
                price = float(price)
                qty = int(qty)
                add = add_product(products, name, price, qty, category)
                print(add if add else f"added: {name}")
            case "restock":
                name, qty = command[1:]
                if not qty.isdigit() or int(qty) <= 0:
                    print(f"invalid: {name}")
                    continue
                qty = int(qty)
                restock_product = restock(products, name, qty)
                print(restock_product if restock_product else f"restocked: {name} -> {products[name]['qty']}")
            case "sell":
                name, qty = command[1:]
                if not qty.isdigit() or int(qty) <= 0:
                    print(f"invalid: {name}")
                    continue
                qty = int(qty)
                sell_product = sell(products, name, qty)
                print(sell_product if sell_product else f"sold: {name}, left {products[name]['qty']}")
            case "report":
                report(products)
            case "quit":
                break
            case _:
                break
main()
````
