## H2 — FizzBuzz Data

### The labelling rule (classic FizzBuzz)

For a number `n`:

- if `n` is divisible by **3** → label is `"fizz"`
- if `n` is divisible by **5** → label is `"buzz"`
- if `n` is divisible by **both 3 and 5** → label is `"fizzbuzz"`
- if `n` is divisible by **neither** → label is the number itself

So `3` → fizz, `5` → buzz, `15` → fizzbuzz, `4` → 4.

### Your task

For each number **1 to 15**, print one line of the form:

`<number>  <even?>  <label>`

where `<even?>` is `True`/`False` and `<label>` follows the rule above.

Example lines:

```
1  False  1
3  False  fizz
5  False  buzz
6  True   fizz
15 True   fizzbuzz
```

### Constraints

No `if` statements. You may use `%`, `==`, `*`, `+`, and `or`.

Hint: `bool` is a subclass of `int` — `True` behaves as `1`, `False` as `0`. What does `True * "fizz"` produce? What does `False * "fizz"` produce?

For the 1–15 repetition: writing 15 lines is fine (Loops is a later lesson) — or use `for n in range(1, 16):` if you already know it.

````python
for i in range(1,16):
    is_even = (i % 2 == 0)
    
    label = (i % 3 == 0) * 'fizz' + (i % 5 == 0) * 'buzz' or str(i)
    
    print(f"{i} {is_even} {label}")
````

# Challenge Project — Profile Builder CLI

Goal: Build a CLI that asks for a person's name, age, height (meters), and is_student (y/n), then prints a profile card.
Output format (acceptance criteria):

```text
=== PROFILE ===
Name: Ada Lovelace
Age: 36 (13140 days old)
Height: 1.70 m (170.0 cm)
Student: True
Types: str, int, float, bool
```
Requirements:
- name → str, age → int, height → float, is_student → bool
- Use type() in the final line
- Use f-string format specs for the cm value
- Run it once with your own data
Stretch: if the user types something other than y/n for student, default to False without crashing.

````python
name = input("What is ur name?")
age = int(input("How old r u?"))
height = float(input("What is ur height in 'meter' not 'american scale'?"))
is_student = True if input("R u a student? y/n") == 'y' else False

print(f'''
=== PROFILE ===
Name: {name}
Age: {age} ({age * 365} days old)
Height: {height:.2f} m ({(height * 100):.2f} cm)
Student: {is_student}
Types: {type(name).__name__}, {type(age).__name__}, {type(height).__name__}, {type(is_student).__name__}
''')
````
