## M2 — Composition Pipeline

**Rule:** Define `clean(text)` (docstring required) that returns the stripped, lowercased text. Define `count_word(text, word)` that uses `clean` and counts occurrences of `word` (case-insensitive).

`text = "  The CAT sat ON the mat  "`. Count the word `"the"`.

**Output format:** exactly `2`

**Constraints:** `count_word` must CALL `clean` (no manual `strip().lower()` inside it). `clean` must have a docstring describing what it does.

````python
# your code here
def clean(text: str):
    """Return stripped and lowered string"""
    return text.strip().lower()

def count_word(text, word):
    return clean(text).count(word.lower())

text = "  The CAT sat ON the mat  "
countword = count_word(text, "the")
print(f"exacly {countword}")
````

## H2 — Refactor Stats into Functions

**Rule:** This flat block computes stats over `scores = [88, 92, 45, 30]`:
```python
total = 0
for s in scores:
    total += s
print(f"mean: {total / len(scores)}")

lowest = scores[0]
for s in scores:
    if s < lowest:
        lowest = s
print(f"min: {lowest}")

highest = scores[0]
for s in scores:
    if s > highest:
        highest = s
print(f"max: {highest}")
```
Refactor it into three functions — `mean(nums)`, `min_of(nums)`, `max_of(nums)` — each doing its own loop and **returning** the value. Call them from the main flow and print.

**Output format:** byte-identical to the flat block:
```
mean: 63.75
min: 30
max: 92
```

**Constraints:** no built-in `sum`/`min`/`max`. No `print` inside the three functions — they return; the caller prints. One responsibility per function.

````python
# your code here

def mean(nums):
    total = 0
    for n in nums:
        total += n
    return total / len(nums)

def min_of(nums):
    lowest = nums[0]
    for n in nums:
        if n < lowest:
            lowest = n
    return lowest

def max_of(nums):
    highest = nums[0]
    for n in nums:
        if n > highest:
            highest = n
    return highest

scores = [88, 92, 45, 30]

print(f"mean: {mean(scores)}")
print(f"min: {min_of(scores)}")
print(f"max: {max_of(scores)}")
````

# Challenge — Refactor the Gradebook into Functions

**Goal:** take your Lesson 4 Gradebook Analyzer and rebuild it as a set of functions — with **byte-identical output**. This is a real refactor: same behavior, better structure.

**Rule (all features REQUIRED):**
1. `validate_size(raw)` — returns a valid class size; invalid (`abc`, `< 1`) → `3` and prints `invalid size -> set to 3`.
2. `read_student(i)` — prompts `Enter name #{i}: ` and `Enter score #{i}: `, applies the guards (blank name → `unknown`, non-digit or out-of-range score → `50`, with the L4 messages), and **returns a `(name, score)` tuple**.
3. `compute_grade(score)` — returns the letter (`A >= 90`, `B >= 80`, `C >= 70`, `D >= 60`, else `F`). No printing.
4. `is_pass(score)` — returns `True`/`False` for `score >= 40`.
5. `build_distribution(grade)` — a small helper that keeps the grade counts; or fold the `dict.get` count into `main`. Your choice, but distribution printing must stay in `A B C D F` order, nonzero counts only.
6. `main()` — orchestrates: get size, loop over students (accumulate total/pass/highest/lowest), print the `=== GRADEBOOK ===` lines with `enumerate`-style numbering, then the `=== SUMMARY ===` and `=== DISTRIBUTION ===` blocks. Keep `main()` short — it delegates, it does not do the math.
7. At least one function carries a docstring.

**Output format** — with input `3 / ada / 88 / ben / 92 / cyn / 45`, byte-identical to Lesson 4:
```
=== GRADEBOOK ===
1. ada - 88 - B - pass
2. ben - 92 - A - pass
3. cyn - 45 - F - pass
=== SUMMARY ===
students: 3
average: 75.00
highest: 92
lowest: 45
pass rate: 100.0%
=== DISTRIBUTION ===
A: 1
B: 1
F: 1
```

**Test cases (all must produce identical output to the L4 version):**
1. The three-student sample above.
2. Size `abc` → `invalid size -> set to 3`.
3. A score of `150` → `invalid score for <name> -> set to 50`.
4. A blank name → `empty name -> set to 'unknown'`.
5. The min/max counterexample — two students scoring `60` and `55` must report `highest: 60` and `lowest: 55` (not 50).
6. A score below 40 → `fail` line and a lower pass rate.


````python
# your code here
def validate_size(raw: str):
    """Validate input is digit and a positive number, return int"""
    if not raw.isdigit() or int(raw) < 1:
        print("invalid size -> set to 3")
        raw = 3
    else:
        raw = int(raw)
    return raw

def read_student(i):
    """Read input user, validate empty name and score, return tuple (name, score)"""
    name = input(f"Enter name #{i}: ") 
    score_input = input(f"Enter score #{i}: ") 
    
    if not name:
        print("empty name -> set to 'unknown'")
        name = "unknown"
    
    if not score_input.isdigit() or not 0 <= int(score_input) <= 100:
        print(f"invalid score for {name} -> set to 50") 
        score = 50
    else:
        score = int(score_input)
    
    return name, score

def compute_grade(score):
    """Mapping the score to a grade, return str grade"""
    if score >= 90:
        grade = "A"
    elif score >= 80:
        grade = "B"
    elif score >= 70:
        grade = "C"
    elif score >= 60:
        grade = "D"
    else:
        grade = "F"
    
    return grade

def is_pass(score):
    """Check status based on score, return bool"""
    return score >= 40

def main():
    class_size = input("Enter class size: ")
    class_size = validate_size(class_size)

    total_score = 0
    total_pass = 0
    highest = None
    lowest = None
    distribution = {}
        
    print("=== GRADEBOOK ===")    
    for i in range(1, class_size+1):
        name, score = read_student(i)
        grade = compute_grade(score)
        status_pass = is_pass(score)
        
        if status_pass:
            total_pass += 1
        total_score += score
        
        if highest is None or score > highest:
            highest = score
        if lowest is None or score < lowest:
            lowest = score
            
        distribution[grade] = distribution.get(grade, 0) + 1
            
        print(f"{i}. {name} - {score} - {grade} - {"pass" if status_pass else "fail"}")
            
    print("=== SUMMARY ===")
    print(f"students: {class_size}")
    print(f"average: {(total_score / class_size):.2f}")
    print(f"highest: {highest}")
    print(f"lowest: {lowest}")
    print(f"pass rate: {(total_pass / class_size * 100):.1f}%")

    print("=== DISTRIBUTION ===")    
    for grade in "ABCDF":
        count = distribution.get(grade, 0)
        if count:
            print(f"{grade}: {count}")
    
main()
````
