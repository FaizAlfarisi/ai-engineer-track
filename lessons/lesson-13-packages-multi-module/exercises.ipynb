## E1 — a module is a file

**Type:** [REAL-WORLD] · easy

**Task:** create `greet.py` with a function `greet(name)` that returns
`"Hello, {name}!"`, then import it and print the result.

**File tree:**
```text
greet.py          <- you create
```

**Constraints:**
- The file is named `greet.py`, so your import name is `greet`.
- `greet()` returns the string; it does not print.
- No `print` line inside `greet.py`.

**Driver:**
```python
from greet import greet
print(greet("ada"))
print(greet("bob") == "Hello, bob!")
```

**Output:**
```
Hello, ada!
True
```

````python
from greet import greet
print(greet("ada"))
print(greet("bob") == "Hello, bob!")
````

## E2 — two files, one imports the other

**Type:** [REAL-WORLD] · easy

**Task:** split a word-count report across two files. `words.py` owns
the counting; `report.py` reuses it.

**File tree:**
```text
words.py          <- you create
report.py         <- you create
```

**Interface contract:**
- `words.py` exposes `count_words(text)` that returns the number of
  words in `text`.
- `report.py` exposes `word_count(text)` that returns the same count
  by importing `count_words` from `words`.

**Constraints:**
- `report.py` must NOT re-implement the counting.
- No top-level `print` in either file.

**Driver:**
```python
from report import word_count
print(word_count("one two two three three three"))
```

**Output:**
```
6
```

````python
from report import word_count
print(word_count("one two two three three three"))
````

```text

    __init__.py
```

- The driver (outside the package) imports absolutely.


**Output:**
```
4 words; longest: longest
```

## Challenge — package a JSON → CSV report tool

**Type:** [REAL-WORLD] · micro build

**Task:** reorganize the report pattern (load → summarize → write)
into a **package**. A `report_tool` package with three modules and one
`main.py` outside it reads `transactions.json`, sums each category,
writes `summary.csv`, and prints its content.

**File tree:**
```text
report_tool/
    __init__.py       <- you create
    load.py           <- you create
    compute.py        <- you create
    write.py          <- you create
main.py               <- you create
transactions.json     <- provided
summary.csv           <- created when the tool runs
```

**Interface contract:**
- `report_tool/load.py` — `load_transactions(path)` reads the JSON
  file and returns the list of records. Missing file or broken JSON
  raise `ValueError`; never an error string as the return value (L9/L11
  rule).
- `report_tool/compute.py` — `summarize(records)` returns a dict
  mapping each `category` to the sum of its `amount`s.
- `report_tool/write.py` — `write_summary(path, totals)` writes a CSV
  with header `category,total`, one row per category sorted
  alphabetically, totals with two decimals, `newline=""` and
  `encoding="utf-8"` (L12 rules).
- `main.py` — `main()` loads, summarizes, writes, then prints the CSV
  content without an added newline, guarded by `__name__ == "__main__"`.
  Imports are absolute (`from report_tool.load import ...`).

**Constraints:**
- Submodules must not print anything themselves.
- Run the driver from the lesson folder so `report_tool` is found.

**Driver:**
```python
from main import main
main()
```

**Output:**
```
category,total
entertainment,60.00
food,153.60
transport,56.00
```

````python
from main import main

main()
````
