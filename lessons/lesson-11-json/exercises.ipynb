## E1 — Round trip with dumps / loads

**Type:** [REAL-WORLD] · easy

**Task:** write `dumps_then_loads(data)` that returns a 2-tuple: the
JSON string and the value read back from it.

**Steps:**
- `text = json.dumps(data)` — the string form.
- `back = json.loads(text)` — the value again.
- Return `(text, back)`.

**Constraints:**
- Use default options for `dumps` (no `indent`, no `separators`).
- Do not print inside the function. The driver prints.
- Paste `def dumps_then_loads(...)` at the top of the cell; the driver
  below stays as shown.

**Driver:**
```python
data = {"name": "alice", "retries": 3, "ok": True, "val": None}
text, back = dumps_then_loads(data)
print(text)
print(back == data)
```

**Output:**
```
{"name": "alice", "retries": 3, "ok": true, "val": null}
True
```


````python
import json

def dumps_then_loads(data):
    text = json.dumps(data)
    back = json.loads(text)
    
    return text, back

data = {"name": "alice", "retries": 3, "ok": True, "val": None}
text, back = dumps_then_loads(data)
print(text)
print(back == data)
````

## E2 — Save and load through a file

**Type:** [REAL-WORLD] · easy

**Task:** write `save_json(path, data)` and `load_json(path)` so a dict or
list survives the trip to disk and back.

**Constraints:**
- Use `Path.write_text` and `Path.read_text` (never raw `open`).
- Pass `encoding="utf-8"` on both.
- `save_json` returns `True`; `load_json` returns the loaded value.
- Do not use `indent` or `separators` here.

**Driver:**
```python
rows = [{"name": "x", "amount": 10.0}, {"name": "y", "amount": 20.5}]
save_json(Path("rows.json"), rows)
back = load_json(Path("rows.json"))
print(back == rows)
print(type(back).__name__)
```

**Output:**
```
True
list
```


````python
from pathlib import Path

def save_json(path: Path, data):
    path.write_text(json.dumps(data), encoding="utf-8")
    return True

def load_json(path: Path):
    return json.loads(path.read_text(encoding="utf-8"))

rows = [{"name": "x", "amount": 10.0}, {"name": "y", "amount": 20.5}]
save_json(Path("rows.json"), rows)
back = load_json(Path("rows.json"))
print(back == rows)
print(type(back).__name__)
````

## E3 — Iterate a loaded list

**Type:** [REAL-WORLD] · easy

**Task:** write `summarize_rows(data)` that returns a 2-tuple: the total
of `n`, and the filtered list of `name` values (`n > 1`).

**Steps:**
- Total: `sum(row["n"] for row in data)`.
- Names: a list comprehension over `data` where `row["n"] > 1`.
- Return `(total, names)`.

**Constraints:**
- Use one list comprehension for the filtered names.
- Use `sum(...)` for the total.
- Do not print inside the function. The driver prints `total` then `names`.
- Paste `def summarize_rows(...)` at the top of the cell; the driver
  below stays as shown.

**Driver:**
```python
data = json.loads('[{"name": "a", "n": 1}, {"name": "b", "n": 2}, {"name": "c", "n": 3}]')
total, names = summarize_rows(data)
print(total)
print(names)
```

**Output:**
```
6
['b', 'c']
```


````python
def summarize_rows(data):
    total = sum(d["n"] for d in data)
    names = [d["name"] for d in data if d["n"] > 1]
    return total, names

data = json.loads('[{"name": "a", "n": 1}, {"name": "b", "n": 2}, {"name": "c", "n": 3}]')
total, names = summarize_rows(data)
print(total)
print(names)
````

## M1 — Safe loader with a default

**Type:** [REAL-WORLD] · medium

**Task:** write `load_settings(path)` that never crashes on a missing or
broken file.

**Constraints:**
- If the file is missing, return `{}`.
- If the file is present but is not valid JSON, return `{}`.
- If it is valid JSON, return the loaded dict.
- Never print, never raise, never return an error string.
- Use `.exists()` (LBYL) for the missing case and catch
  `json.JSONDecodeError` (EAFP) for the broken case.

**Driver:**
```python
Path("good.json").write_text('{"retries": 3}', encoding="utf-8")
Path("broken.json").write_text("{oops", encoding="utf-8")
print(load_settings(Path("missing.json")))
print(load_settings(Path("broken.json")))
print(load_settings(Path("good.json")))
```

**Output:**
```
{}
{}
{'retries': 3}
```


````python
def load_settings(path: Path):
    if not path.exists():
        return {}
    
    try:
        return json.loads(path.read_text(encoding="utf-8"))
    except json.JSONDecodeError:
        return {}

Path("good.json").write_text('{"retries": 3}', encoding="utf-8")
Path("broken.json").write_text("{oops", encoding="utf-8")
print(load_settings(Path("missing.json")))
print(load_settings(Path("broken.json")))
print(load_settings(Path("good.json")))
````

## M3 — Typed loader with a clean re-raise, and a match route

**Type:** [REAL-WORLD] · medium

**Task:** write `load_typed(path)` and `describe(record)`.

`load_typed` reads `key=value` lines and converts each value to a typed
field. It must fail cleanly: when a value cannot be converted, it raises
`ValueError(f"bad setting: {key}")` **with `from None`** so the caller sees
only your message — no chained traceback wall (this is the exception-detail
point from Quiz 01 D2).

**Constraints:**
- `SPEC = {"retries": int, "rate": float, "name": str}` at the top.
- Skip blank lines and unknown keys.
- When `SPEC[key]` fails, raise `ValueError(f"bad setting: {key}") from None`.
- `describe(record)` uses `match` (mapping patterns): an income record
  returns `"+<amount>"`, an expense record returns `"-<amount>"`, anything
  else returns `"?"`.

**Driver:**
```python
cfg = Path("app.cfg")
cfg.write_text("retries=3\nrate=12.5\nname=alice\n", encoding="utf-8")
print(load_typed(cfg))
bad = Path("bad.cfg")
bad.write_text("retries=oops\n", encoding="utf-8")
try:
    load_typed(bad)
except ValueError as e:
    print("ValueError:", e)
records = [{"kind": "income", "amount": 5}, {"kind": "expense", "amount": 3}, {"kind": "junk", "amount": 1}]
print([describe(r) for r in records])
```

**Output:**
```
{'retries': 3, 'rate': 12.5, 'name': 'alice'}
ValueError: bad setting: retries
['+5', '-3', '?']
```

Example: a line with no `=` (for example `justtext`) splits to an empty
value via `partition`, so `int("")` fails and the loader raises a clean
`ValueError`. That is intended.


````python
SPEC = {
    "retries": int,
    "rate": float,
    "name": str
}

def load_typed(path: Path):
    result = {}
    
    lines = path.read_text(encoding="utf-8")
    for line in lines.splitlines():
        key, value = line.split("=")
        
        if key in SPEC:
            try:
                result[key] = SPEC[key](value)
            except ValueError:
                raise ValueError(f"bad setting: {key}") from None
    return result

def describe(record):
    match record:
        case {"kind": "income", "amount": amt}:
            return f"+{amt}"
        case {"kind": "expense", "amount": amt}:
            return f"-{amt}"
        case _:
            return "?"
    return


cfg = Path("app.cfg")
cfg.write_text("retries=3\nrate=12.5\nname=alice\n", encoding="utf-8")
print(load_typed(cfg))
bad = Path("bad.cfg")
bad.write_text("retries=oops\n", encoding="utf-8")
try:
    load_typed(bad)
except ValueError as e:
    print("ValueError:", e)
records = [{"kind": "income", "amount": 5}, {"kind": "expense", "amount": 3}, {"kind": "junk", "amount": 1}]
print([describe(r) for r in records])
````

## H1 — Debug: bugs in a JSON folder tool

**Type:** [REAL-WORLD] · debug drill

`collect_json(folder)` should read every `.json` **file** in a folder and

**Observed as-is:** an empty list and `False`. Then, step by step, other

**Rules:**

**The bug families (four):**

2. `iterdir()` yields folders too — the driver makes a folder named
   `config.json`, and reading it crashes. Guard it.
3. The result is the wrong shape: it should be ONE merged dict, not a

4. (the fourth is a consequence of the third — find it as you go.)

**Hints (ask one at a time):**

- A folder is a child too — what test does the loop need?
- What does `result.append(...)` build? What should a merge build instead?

**Driver:**
```python
folder = Path("conf")
folder.mkdir(exist_ok=True)
(folder / "app.json").write_text('{"retries": 3}', encoding="utf-8")
(folder / "db.json").write_text('{"host": "x"}', encoding="utf-8")
(folder / "config.json").mkdir()
expected = {"retries": 3, "host": "x"}
print(collect_json(folder))
print(collect_json(folder) == expected)
```


````python
from pathlib import Path
import json

def collect_json(folder):
    result = {}
    for child in folder.iterdir():
        if not child.is_file():
            continue
        if child.suffix == ".json":
            text = json.loads(child.read_text(encoding="utf-8"))
            for key, value in text.items():
                result[key] = value
    return result


# --- driver (do not change) ---
folder = Path("conf")
folder.mkdir(exist_ok=True)
(folder / "app.json").write_text('{"retries": 3}', encoding="utf-8")
(folder / "db.json").write_text('{"host": "x"}', encoding="utf-8")
(folder / "config.json").mkdir(exist_ok=True)
expected = {"retries": 3, "host": "x"}
print(collect_json(folder))
print(collect_json(folder) == expected)
````

## H2 — Interview drill: count by key with edge cases

**Type:** [INTERVIEW]

**Task:** write `count_by(items, key)` that returns a dict: each distinct
value of `key` mapped to how many items have it.

**Constraints:**
- Use `it.get(key, "missing")` for an item without that key — that value
  is counted as `"missing"`.
- Skip items whose value is `None` (do not count them).
- Preserve the first-seen insertion order of values.

**Driver:**
```python
records = [
    {"kind": "income", "tag": "a"},
    {"kind": "expense", "tag": "b"},
    {"kind": "income", "tag": "a"},
    {"kind": "expense", "tag": None},
    {"amount": 5},
]
print(count_by(records, "tag"))
print(count_by(records, "kind"))
```

**Output:**
```
{'a': 2, 'b': 1, 'missing': 1}
{'income': 2, 'expense': 2, 'missing': 1}
```


````python
def count_by(items, key):
    result = {}
    
    for item in items:
        value = item.get(key, "missing")
        
        if value is None:
            continue
        
        result[value] = result.get(value, 0) + 1
        
    return result

records = [
    {"kind": "income", "tag": "a"},
    {"kind": "expense", "tag": "b"},
    {"kind": "income", "tag": "a"},
    {"kind": "expense", "tag": None},
    {"amount": 5},
]
print(count_by(records, "tag"))
print(count_by(records, "kind"))
````

## Challenge — Expense JSON seed

**Type:** [REAL-WORLD] · micro build

Build a tiny, re-runnable Expense seed — the first stone of the Stage 2
Expense Manager.

**Functions to write:**

`load(path) -> list`
- Missing file: return `[]`.
- Present but broken JSON: return `[]`.
- Present and a JSON **list**: return it.
- Present and NOT a list (e.g. a dict): return `[]`.
- Never print, never raise, never return an error string.

`categorize(record) -> float`
- Use `match` (mapping patterns):
  - `{"kind": "income", "amount": amt}` → `amt`
  - `{"kind": "expense", "amount": amt}` → `-amt`
  - anything else → `0`

`summarize(spends) -> dict`
- Map each `category` to the **signed total** of its amounts (income adds,
  expense subtracts), using `categorize`.
- Start each total at `0.0` so the result is deterministic.

`save(path, spends) -> bool`
- Write `json.dumps(spends, indent=2)` with `write_text`, `encoding="utf-8"`.
- Return `True`.

**Constraints:**
- Every read/write passes `encoding="utf-8"`.
- File I/O goes through pathlib (`Path`), never raw `open`.
- Loaders return data or a default — never an error string.

**Driver:**
```python
p = Path("expenses.json")
print("missing:", load(Path("nope.json")))
p.write_text(json.dumps([
    {"category": "food", "amount": 12.5, "kind": "expense"},
    {"category": "travel", "amount": 40.0, "kind": "expense"},
    {"category": "salary", "amount": 2000.0, "kind": "income"},
], indent=2), encoding="utf-8")
spends = load(p)
print("summarize:", summarize(spends))
save(Path("saved.json"), spends)
print("roundtrip:", load(Path("saved.json")) == spends)
print("saved line1:", repr(Path("saved.json").read_text(encoding="utf-8").splitlines()[1]))
bad = Path("broken.json")
bad.write_text("{oops", encoding="utf-8")
print("broken:", load(bad))
```

**Output:**
```
missing: []
summarize: {'food': -12.5, 'travel': -40.0, 'salary': 2000.0}
roundtrip: True
saved line1: '  {'
broken: []
```


````python
def load(path: Path):
	if not path.exists():
		return []
	
	try:
		text = json.loads(path.read_text(encoding="utf-8"))
	except json.JSONDecodeError:
		return []
	
	if type(text).__name__ != "list":
		return []

	return text

def categorize(record):
	match record:
		case {"kind": "income", "amount":amt}:
			return amt
		case {"kind": "expense", "amount":amt}:
			return -amt
		case _:
			return 0.0

def summarize(spends):
	result = {}
	for spend in spends:
		result[spend['category']] = categorize(spend)
	return result

def save(path: Path, spends):
	path.write_text(json.dumps(spends, indent=2), encoding="utf-8")
	return True

p = Path("expenses.json")
print("missing:", load(Path("nope.json")))
p.write_text(json.dumps([
    {"category": "food", "amount": 12.5, "kind": "expense"},
    {"category": "travel", "amount": 40.0, "kind": "expense"},
    {"category": "salary", "amount": 2000.0, "kind": "income"},
], indent=2), encoding="utf-8")
spends = load(p)
print("summarize:", summarize(spends))
save(Path("saved.json"), spends)
print("roundtrip:", load(Path("saved.json")) == spends)
print("saved line1:", repr(Path("saved.json").read_text(encoding="utf-8").splitlines()[1]))
bad = Path("broken.json")
bad.write_text("{oops", encoding="utf-8")
print("broken:", load(bad))
````
