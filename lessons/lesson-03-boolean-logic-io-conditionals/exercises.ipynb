## E1 — Truthiness Predictor

**Rule:** For each value below, write your predicted truthiness as a `#` comment FIRST, then compute it in code.

**Values:** `0`, `0.0`, `""`, `[]`, `"0"`, `" "`, `[0]`, `None`, `False`

**Output format:** one line per value, exactly `repr(value) -> bool(value)` (e.g. `0 -> False`).

**Constraints:** use `bool()`, no `if`. Use the `!r` format spec to get the repr (the quotes around strings).

````python
# 0 -> False
print(f"{0!r} -> {bool(0)}")

# 0.0 -> False
print(f"{0.0!r} -> {bool(0.0)}")

# "" -> False
print(f"{""!r} -> {bool("")}")

# "0" -> True
print(f"{"0"!r} -> {bool("0")}")

# [] -> False
print(f"{[]!r} -> {bool([])}")

# " " -> True
print(f"{" "!r} -> {bool(" ")}")

# [0] -> True
print(f"{[0]!r} -> {bool([0])}")

# None -> False
print(f"{None!r} -> {bool(None)}")

# False -> False
print(f"{False!r} -> {bool(False)}")
````

## E4 — Name Validator

**Rule:** Read a name (strip it). If empty → reject; else if shorter than 2 characters → reject; else greet.

**Output format:**
```
invalid: empty name
invalid: too short
ok: Hello, Ada
```

**Constraints:** test the empty check with truthiness (`if not name:` — NOT `name == ""`). Length via `len(name)`. Professional prompt.

````python
# your code here

test_names = ["", "ue", "faiz"]
for name in test_names:
    if not name:
        print("invalid: empty name")
    elif len(name) < 2:
        print("invalid: too short")
    else:
        print(f"ok: Hello, {name}")
````

## E5 — Command Dispatcher (first `match`)

**Rule:** For each command in `["start", "stop", "pause", "start"]`, dispatch it with `match`.

**Output format:** `start -> starting`, `stop -> stopping`, anything else `pause -> unknown command`.

**Constraints:** one `match`; `case "start":`, `case "stop":`, and a `_` wildcard. Loop over the list allowed.

````python
# your code here
commands = ["start", "stop", "pause", "start"]

for command in commands:
    match command:
        case "start": 
            print(f"{command} -> starting")
        case "stop": 
            print(f"{command} -> stopping")
        case _: 
            print(f"{command} -> unknown command")

        
````

## M1 — Grade Ladder

**Rule:** Read a score (`0`–`100`). Guard clause first: if outside that range, print `invalid score` and stop. Otherwise classify: `A >= 90`, `B >= 80`, `C >= 70`, `D >= 60`, `F < 60`.

**Output format:** `Score: 85 -> grade B` and `invalid score` for out-of-range.

**Constraints:** one `if`/`elif`/`else` ladder; the range guard must be the FIRST condition, ordered from most specific to most general. Test `101`, `100`, `85`, `60`, `59`.

````python
# your code here
scores = [101, 100, 85, 60, 59]

for score in scores:
    if not (0 <= score <= 100):
        print("invalid score")
        continue
    elif score >= 90:
        grade = "A"
    elif score >= 80:
        grade = "B"
    elif score >= 70:
        grade = "C"
    elif score >= 60:
        grade = "D"
    else:
        grade = "F"
    print(f"Score: {score} -> grade {grade}")
````

## M2 — Discount Rules

**Rule:** Decide a discount for each `(member, spend, referred)` tuple:
- member AND (`spend >= 100` OR `referred`) → `20%`
- NOT member AND `spend >= 200` → `10%`
- otherwise → `0%`

**Cases:** `(True, 120.0, False)`, `(True, 50.0, True)`, `(True, 50.0, False)`, `(False, 250.0, False)`, `(False, 100.0, False)`

**Output format:** `member=True spend=120.0 referred=False -> 20%` (one line per case).

**Constraints:** one compound condition per branch, using `and`, `or`, `not`, and parentheses. Loop over the cases.

````python
# your code here
cases = [(True, 120.0, False), (True, 50.0, True), (True, 50.0, False), (False, 250.0, False), (False, 100.0, False)]

for case in cases:
    member, spend, referred = case
    if member and (spend >= 100 or referred):
        discount = 20
    elif not member and spend >= 200:
        discount = 10
    else:
        discount = 0
    print(f"member={member} spend={spend} referred={referred} -> {discount}%")
````

## H2 — Escalation Router (`match` + guard)

**Rule:** Route each `(severity, count)` entry. Use `match` with a GUARD for the escalation, `|` OR-patterns, and a `_` wildcard.

**Entries:** `("INFO", 5)`, `("WARNING", 2)`, `("ERROR", 3)`, `("CRITICAL", 9)`, `("DEBUG", 0)`, `("UNKNOWN", 1)`

**Routing rules (first match wins):**
1. `CRITICAL` OR `ERROR` AND `count >= 5` → `alert on-call`
2. `CRITICAL` OR `ERROR` → `log to error file`
3. `WARNING` → `log to warning file`
4. `INFO` OR `DEBUG` → `log to info file`
5. anything else → `drop`

**Output format:** `CRITICAL(9) -> alert on-call` (one line per entry).

**Constraints:** the guard must be `if count >= 5` on the first case; use `|` at least once; `_` last.

````python
# your code here
entries = [("INFO", 5), ("WARNING", 2), ("ERROR", 3), ("CRITICAL", 9), ("DEBUG", 0), ("UNKNOWN", 1)]

for severity, count in entries:
    match severity:
        case "CRITICAL" | "ERROR" if count >= 5:
            message = "alert on-call" 
        case "CRITICAL" | "ERROR":
            message = "log to error file"
        case "WARNING":
            message = "log to warning file"
        case "INFO" | "DEBUG":
            message = "log to info file"
        case _:
            message = "drop"
    
    print(f"{severity}({count}) -> {message}")
````

# Challenge — Support-Ticket Router CLI

**Goal:** route support tickets to departments through two connected stages.

**Stage 1 — Validation (`if`/`elif`/`else` + compound conditions).** Read three inputs:
- `priority` (`low`/`medium`/`high`/`critical`) — case-insensitive. If invalid → print `invalid priority -> set to 'low'` and use `low`.
- `category` — if blank → print `empty category -> set to 'general'` and use `general`.
- `agent load` (`0`–`10` int) — if not a digit or out of range → print `invalid load -> set to 5` and use `5`. (Hint: `"7".isdigit()` — string methods are your friend.)

**Stage 2 — Dispatch (`match`).** Route:
- `critical` AND `load >= 8` → `ON-CALL TEAM` (use a **match guard**)
- else by category: `login` | `auth` → `auth team` · `billing` | `payment` → `finance team` · `crash` | `bug` → `engineering team` · anything else → `general team`

**Output format (section headers exact):**
```
=== TICKET VALIDATION ===
priority: high
category: login
load: 7
=== ROUTING ===
team: auth team
```

Validation messages print before the summary.

**Test these four cases:**
1. `CRITICAL` / `login` / `8` → ON-CALL TEAM
2. `urgent` / `(blank)` / `99` → all three defaults, then `general team`
3. `high` / `billing` / `4` → `finance team`
4. `low` / `crash` / `10` → `engineering team`


````python
# your code here

test_cases = [("CRITICAL", "login", "8"), ("urgent", "", "99"), ("high", "billing", "4"), ("low", "crash", "10")]
priorities = ["low", "medium", "high", "critical"]

for priority, category, agent_load in test_cases:
    priority = priority.strip().lower()
    category = category.strip().lower()
    # stage 1 
    print("=== TICKET VALIDATION ===")
    if priority not in priorities:
        priority = "low"
        print("invalid priority -> set to 'low'")
    if not category:
        print("empty category -> set to 'general'")
        category = "general"
    if not agent_load.isdigit() or not 0 <= int(agent_load) <= 10:
        print("invalid load -> set to 5")
        agent_load = 5
    else:
        agent_load = int(agent_load)

    # stage 2
    match (priority, category):
        case ("critical", _) if agent_load >= 8:
            routing_message = "ON-CALL TEAM"
        case (_, "login" | "auth"):
            routing_message = "auth team"
        case (_ , "billing" | "payment"):
            routing_message = "finance team"
        case (_ , "crash" | "bug"):
            routing_message = "engineering team"
        case (_ , _):
            routing_message = "general team"

    print(f"priority: {priority}")
    print(f"category: {category}")
    print(f"load: {agent_load}")
    print("=== ROUTING ===")
    print(f"team: {routing_message}")

````
