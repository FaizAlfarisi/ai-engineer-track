## E4 — annotations never enforce

**Type:** [REAL-WORLD] · easy

**Task:** create `warn.py` with one function:

- `score_bonus(score)` returns `score + 1` — annotated `score: int`,
  return `int`.

This is a trap. The driver calls `score_bonus` with a float and then
with a string on purpose. Annotations are labels, not guards: the float
slips through and returns `4.0`. The string never reaches the
annotation — Python's `+` operator itself raises. Read the Output block
first, predict it, then run.

**Constraints:**
- The function itself is annotated honestly. Do not `fix` the driver
  — the wrong-typed calls are the exercise.
- No `try/except` inside `warn.py` — the driver holds the guard.

**File tree:**
```text
warn.py            <- you create
```

**Driver:**
```python
from typing import get_type_hints
import warn

print(get_type_hints(warn.score_bonus))
print(warn.score_bonus(1))
print(warn.score_bonus(3.0))
try:
    warn.score_bonus("5")
except TypeError as e:
    print(type(e).__name__)
```

**Output:**
```
{'score': <class 'int'>, 'return': <class 'int'>}
2
4.0
TypeError
```

````python
from typing import get_type_hints
import warn

print(get_type_hints(warn.score_bonus))
print(warn.score_bonus(1))
print(warn.score_bonus(3.0))
try:
    warn.score_bonus("5")
except TypeError as e:
    print(type(e).__name__)
````
