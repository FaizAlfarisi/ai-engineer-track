## M1 — composition: a `Ledger` that holds `Transaction`s

**Type:** [REAL-WORLD] · medium

**Task:** create `ledger.py` with one class that reuses your E1 file:

- `class Ledger` — `__init__(self, owner: str) -> None`; attributes
  `owner` (`str`) and `transactions` (an empty `list[Transaction]`).
- Method `add(self, txn: Transaction) -> None` — appends to the list.
- Method `net(self) -> float` — sum of every `txn.signed_amount()`.
- Method `report(self) -> str` — a multi-line string: first line
  `f"{owner}: {net:.2f}"`, then one line per transaction **sorted by
  date** ascending, formatted as `f"{date} {category} {signed_amount:+.2f}"`
  (the `+` shows income as `+100.00`, expense as `-12.50`).

Composition: `Ledger` **has** `Transaction`s. This is the pattern that
keeps a whole object graph together instead of scattering dicts
through function calls.

**Constraints:**
- Import the `Transaction` class from your E1 file:
  `from transaction import Transaction`.
- Methods annotated; `add` returns `None`; the list is typed
  `list[Transaction]`.
- `report` returns the string — it does not print.

**File tree:**
```text
ledger.py           <- you create
transaction.py      <- from E1 (already created)
```

**Driver:**
```python
from ledger import Ledger
from transaction import Transaction

book = Ledger("ada")
book.add(Transaction("2026-09-04", "food", "expense", 12.5))
book.add(Transaction("2026-09-25", "salary", "income", 100.0))
book.add(Transaction("2026-10-01", "food", "expense", 8.0))

print(book.owner)
print(book.net())
print(book.report())
```

**Output:**
```
ada
79.5
ada: 79.50
2026-09-04 food -12.50
2026-09-25 salary +100.00
2026-10-01 food -8.00
```

````python
from ledger import Ledger
from transaction import Transaction

book = Ledger("ada")
book.add(Transaction("2026-09-04", "food", "expense", 12.5))
book.add(Transaction("2026-09-25", "salary", "income", 100.0))
book.add(Transaction("2026-10-01", "food", "expense", 8.0))

print(book.owner)
print(book.net())
print(book.report())
````

## M2 — composition + delegation: a `Session` wrapping a `Ledger`

**Type:** [REAL-WORLD] · medium

**Task:** create `session.py` with one class that composes your M1 work:

- `class Session` — `__init__(self, ledger: Ledger, path: Path) -> None`;
  stores `ledger` and `path`.
- Method `add(self, date: str, category: str, kind: str, amount: float) -> None`
  — builds a `Transaction` and **delegates** to `self.ledger.add(...)`.
- Method `net(self) -> float` — delegates to `self.ledger.net()`.
- Method `save(self) -> None` — writes `{"owner": ... , "net": ...}`
  to `self.path` as JSON (Lesson 11: `json.dump`, `indent=2`). The
  driver reads it back with the Lesson 11 loader shape.

Delegation: `Session` does not reimplement anything — it forwards calls
to the object it holds. That is composition with a thin outer layer.

**Constraints:**
- `from ledger import Ledger` and `from transaction import Transaction`.
- All methods annotated; mutators return `None`.
- The driver creates `session_out.json` in the lesson folder — that is
  expected; delete it before committing.

**File tree:**
```text
session.py          <- you create
ledger.py           <- from M1 (already created)
transaction.py      <- from E1 (already created)
```

**Driver:**
```python
import json
from pathlib import Path
from ledger import Ledger
from session import Session

book = Ledger("ada")
sess = Session(book, Path("session_out.json"))

sess.add("2026-09-04", "food", "expense", 12.5)
sess.add("2026-09-25", "salary", "income", 100.0)
print(sess.net())
sess.save()
print(json.loads(Path("session_out.json").read_text(encoding="utf-8")))
```

**Output:**
```

{'owner': 'ada', 'net': 87.5}
```

````python
import json
from pathlib import Path
from ledger import Ledger
from session import Session

book = Ledger("ada")
sess = Session(book, Path("session_out.json"))

sess.add("2026-09-04", "food", "expense", 12.5)
sess.add("2026-09-25", "salary", "income", 100.0)
print(sess.net())
sess.save()
print(json.loads(Path("session_out.json").read_text(encoding="utf-8")))
````
