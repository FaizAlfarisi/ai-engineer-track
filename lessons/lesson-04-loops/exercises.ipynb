## E1 — Words & Lengths

**Rule:** For each word in `["cat", "window", "defenestrate"]`, print the word and its length.

**Output format:** `word: length` (one line per word).

**Constraints:** a `for` loop over the list — iterate ITEMS, not `range(len(words))`. Use `len()` inside the loop.

## E2 — Sum Accumulator

**Rule:** With `n = 10`, compute the sum of `1 + 2 + ... + n` using an accumulator.

**Output format:** `sum 1..10 = 55`

**Constraints:** `total = 0` BEFORE the loop, `range(1, n + 1)`, and `total += n`. Do NOT use the built-in `sum()` — build it by hand.

````python
total = 0
for n in range(1, 11):
    total += n
print(f"sum 1..{n} = {total}")
````

## M2 — Fold-Once Frequency

**Rule:** Count occurrences of each word in `words = ["Data", "data", "is", "IS", "gold"]`, case-insensitively, printing in FIRST-SEEN order.

**Output format:**
```
data: 2
is: 2
gold: 1
```

**Constraints:** the fold-once rule — build ONE lowercased copy of the list and reuse it (do NOT call `.lower()` inside the loop). Use a dict.

````python
# your code here
words = ["Data", "data", "is", "IS", "gold"]
normalized_word = [word.lower() for word in words]
# normalized_word = list(map(str.lower, words))

word_counts = {}
for word in normalized_word:
    word_counts[word] = word_counts.get(word, 0) + 1
    
print("\n".join([f"{word}: {count}" for word, count in word_counts.items()]))
````

## M3 — Search (loop `else`)

**Rule:** Search `items = ["ada", "ben", "cyn"]` for `"ben"` and then for `"zoe"`, using the loop-`else` idiom.

**Output format:**
```
found: ben at 1
not found: zoe
```

**Constraints:** `for`/`else` with `break` — NO flag variable. Use `enumerate()` for the index.

````python
items = ["ada", "ben", "cyn"]

# Path 1
target_1 = "ben"
for i, item in enumerate(items):
    if item == target_1:
        print(f"found: {item} at {i}")
        break
else:
    print(f"not found: {target_1}")

# Path 2
target_2 = "zoe"
for i, item in enumerate(items):
    if item == target_2:
        print(f"found: {item} at {i}")
        break
else:
    print(f"not found: {target_2}")
````

**Rule (part b):** Build a dict comprehension mapping each word in `["cat", "window"]` to its length.

**Output format:** `{'cat': 3, 'window': 6}`

**Constraints:** the comprehension must be byte-identical to the loop; prove it with `==`.

````python
# your code here
result = []
for n in range(1, 11):
    if n % 2 == 0:
        result.append(n ** 2)
print(result)

result_comp = [n**2 for n in range(1, 11) if n%2 == 0]
print(result_comp)
print(f"equal: {bool(result == result_comp)}")

names = ["cat", "window"]

name_counts = {name: len(name) for name in names}
print(name_counts)
````

# Challenge — Gradebook Analyzer CLI

**Goal:** build a loop-driven class grade report.

**Rule (all features REQUIRED):**
1. Prompt `Enter class size:` — if not a digit or `< 1`, print `invalid size -> set to 3` and use `3`.
2. Loop over the class size with a `for` loop. Per student, prompt `Enter name #i:` and `Enter score #i:`.
3. Name blank → print `empty name -> set to 'unknown'` and use `unknown`. Score not a digit or outside `0`–`100` → print `invalid score for <name> -> set to 50` and use `50`.
4. Grade each score: `A >= 90`, `B >= 80`, `C >= 70`, `D >= 60`, else `F`. Pass = `score >= 40`.
5. Accumulate total, lowest, highest. Compute average (2 decimals) and pass rate (1 decimal).
6. Build a grade distribution dict, then print nonzero counts in `A B C D F` order.

**Output format (headers exact):**
```
=== GRADEBOOK ===
1. ada - 88 - B - pass
2. ben - 92 - A - pass
3. cyn - 45 - F - pass
=== SUMMARY ===
students: 3
average: 75.00
highest: 92
lowest: 45
pass rate: 100.0%
=== DISTRIBUTION ===
A: 1
B: 1
F: 1
```

The per-student `1. ada - 88 - B - pass` lines use `enumerate` over the stored names/scores.

**Test cases:**
1. The three-student run above — verify format exactly.
2. Size `abc` → `invalid size -> set to 3`.
3. A score of `150` → `invalid score for ... -> set to 50`.
4. A blank name → `empty name -> set to 'unknown'`.
5. A score below 40 → `fail` in the GRADEBOOK line and a lower pass rate.


````python
class_size = input("Enter class size: ")
if not class_size.isdigit() or int(class_size) < 1:
    print("invalid size -> set to 3")
    class_size = 3
else:
    class_size = int(class_size)

total_score = 0
total_pass = 0
highest = None
lowest = None
distribution = {}
    
print("=== GRADEBOOK ===")    
    
for i in range(1, class_size+1):
    name = input(f"Enter name #{i}: ") 
    score_input = input(f"Enter score #{i}: ") 
    
    if not name:
        print("empty name -> set to 'unknown'")
        name = "unknown"
    
    if not score_input.isdigit() or not 0 <= int(score_input) <= 100:
        print(f"invalid score for {name} -> set to 50") 
        score = 50
    else:
        score = int(score_input)
        
    total_score += score
    
    if highest is None or score > highest:
        highest = score
    if lowest is None or score < lowest:
        lowest = score
        
    if score >= 90:
        grade = "A"
    elif score >= 80:
        grade = "B"
    elif score >= 70:
        grade = "C"
    elif score >= 60:
        grade = "D"
    else:
        grade = "F"
    
    if score >= 40:
        status = "pass"
        total_pass += 1 
    else:
        status = "fail"
        
    distribution[grade] = distribution.get(grade, 0) + 1
        
    print(f"{i}. {name} - {score} - {grade} - {status}")
        
print("=== SUMMARY ===")
print(f"students: {class_size}")
print(f"average: {(total_score / class_size):.2f}")
print(f"highest: {highest}")
print(f"lowest: {lowest}")
print(f"pass rate: {(total_pass / class_size * 100):.1f}%")

print("=== DISTRIBUTION ===")    
for grade in "ABCDF":
    print(f"{grade}: {distribution.get(grade, 0)}")
````
