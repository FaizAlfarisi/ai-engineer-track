## M1 — Reproducibility

**Task:** with `random.seed(1)`, build `a` = five `randint(1, 100)` values
via a list comprehension. Re-seed with `1`, build `b` identically. Print
`a == b` and `a`.

**Output:**
```
True
[18, 73, 98, 9, 33]
```

**Constraint:** two separate `seed(1)` calls — one before each build — and
both comprehensions must be identical expressions. Keywords: **same seed,
same sequence**, **reproducibility**.

````python
import random

random.seed(1)
a = [random.randint(1,100) for _ in range(5)]

random.seed(1)
b = [random.randint(1,100) for _ in range(5)]

print(a == b)
print(a)
````

## M2 — Coin flips with `dict.get`

**Task:** `random.seed(42)`. Flip a coin (`random.choice(["H", "T"])`)
100 times, counting results into a dict with `dict.get`. Print the dict
sorted by key.

**Output:**
```
{'H': 56, 'T': 44}
```

**Constraint:** count with `counts[f] = counts.get(f, 0) + 1` — no `if`
branch. Print `dict(sorted(counts.items()))`. Keywords: **accumulate**,
**fold once**.

````python
import random

random.seed(42)
counts = {}

for _ in range(100):
    f = random.choice(["H", "T"])
    counts[f] = counts.get(f, 0) + 1
    
print(dict(sorted(counts.items())))
````

## H2 — Interview drill: coupon collector

A fair die is rolled until **every face 1–6 has appeared at least once**.
`random.seed(5)`. Count the rolls and print the total and the set of faces
seen.

**Output:**
```
11 rolls to collect all 6 faces
seen: [1, 2, 3, 4, 5, 6]
```

**Constraints (interview-grade):**
- One `while` loop; loop while `len(seen) < 6` (keyword: **coupon
  collector**, **birthday/coverage problem**).
- Track seen faces with a **set**; add the roll each iteration, increment
  a roll counter.
- `print(rolls, "rolls to collect all 6 faces")` then
  `print("seen:", sorted(seen))`.
- `seed(5)` exactly once, before the loop.

````python
import random

seen = set()
rolls = 0
random.seed(5)

while len(seen) < 6:
    seen.add(random.choice([1,2,3,4,5,6]))
    rolls += 1
        
print(f"{rolls} rolls to collect all 6 faces")
print(f"seen: {sorted(seen)}")
````

## Challenge — Random Team Picker CLI

Build a **command-line** program (run with `python solution.py ...`) that
picks a random team of a given size from a list of names. This is the first
CLI that reads real command-line arguments.

### Usage

```
python solution.py <count> <name> [name ...]
python solution.py --seed 5 2 ada ben cyn
```

### Behavior

- `<count>` must be a positive integer no larger than the number of names.
- Names are the remaining arguments.
- Optional `--seed N` (must come first if present) makes the pick
  reproducible via `random.seed(N)`.
- Invalid count (non-int, < 1, or > len(names)) → print
  `invalid count: N` and exit.
- Fewer than 2 arguments → print the usage line and exit.

### Output format

```
=== winners ===
1. cyn
2. ben
```

Winners are `random.sample(names, count)` (no repeats), numbered 1-based.

### Example

```
python solution.py --seed 5 2 ada ben cyn
=== winners ===
1. cyn
2. ben
```

### Structure constraints (the real test)

```python
import random
import sys

def parse_args(argv): ...       # -> (seed, count, names); None seed if absent
def pick(names, count): ...     # -> list via random.sample, one line
def print_winners(winners): ... # -> prints header + numbered winners
def main(): ...                 # -> parses argv, validates, picks, prints

if __name__ == "__main__":
    main()
```

- `parse_args` slices `argv[1:]`; a leading `--seed` consumes the next arg
  as the seed (int).
- Validation happens in `main()` *before* calling `pick`.
- `main()` reads nothing — it takes `sys.argv` (passed from the guard).
- `pick` must call `random.seed(seed)` first **if** a seed was given, then
  `return random.sample(names, count)`.
- `print_winners` prints `=== winners ===` then `f"{i}. {name}"` per
  winner.

Verify in a terminal: the seeded example above must produce exactly the
shown output, and these must fail gracefully (usage / invalid count, no
crash): `python solution.py`, `python solution.py 0 ada`, `python
solution.py 5 ada ben`.

````python
import random 
import sys

def parse_args(argv):
    count = 0
    names = []
    seed = None
    
    if "--seed" == argv[0]:
        seed = int(argv[1]) # seed value is set after --seed, so that it is the second index in argv
        count = int(argv[2])
        names = argv[3:]
    else:
        count = int(argv[0])
        names = argv[1:]
        
    return names, count, seed

def pick(names, count, seed=None):
    if seed is not None:
        random.seed(seed)
        
    return random.sample(names, count)

def print_winners(winners):
    print("=== winners ===")
    print('\n'.join(f"{i+1}. {winner}" for i, winner in enumerate(winners)))

def main():
    argv = sys.argv
    
    if len(argv) < 3:
        print("the argv is required to be at least lenghted 3")
        return 
    
    names, count, seed = parse_args(argv[1:])
    
    if count < 1 or count > len(names):
        print("cant count less than 1 or bigger than total names")
        return
    
    winners = pick(names, count, seed)
    print_winners(winners)
    return

if __name__ == "__main__":
    main()
````
