```python
def harvest_count(stream):
    it = iter(stream)
    total = sum(1 for _ in it)
    got = list(it)
    return total, got

print(harvest_count(["a", "b", "c"]))
print(harvest_count(iter(["a", "b", "c"])))
```

```python
pulled = []

def metered():
    for x in [18.5, 42.0, 19.0]:
        pulled.append(x)
        yield x

hot = [x for x in metered() if x > 40.0]
print(pulled)
print(hot)
print(next(hot))
print(pulled)
print(list(hot))
```

```python
def usable(name):
    return not isinstance(name, str) and len(name.strip()) > 0

print(usable("  data.csv "))
print(usable(""))
print(usable(None))
print(usable(123))
```

```python
THRESHOLD = 40.0

def alerts(readings):
    """Yield readings strictly over THRESHOLD."""
    found = (r for r in readings if r > THRESHOLD)
    yield found
    
list(alerts([18.0, 45.0, 20.0]))
```
